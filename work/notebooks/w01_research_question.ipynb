## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

Before building a complex model, I wanted to see if there is actually a problem to solve. I loaded the starter data to check the distribution of CTR and see how many pages might be underperforming. The numbers below show that a significant portion of pages have low engagement, making a ranking system highly valuable.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Paste your Raw GitHub URL here inside the quotes
url = "https://raw.githubusercontent.com/Mxnaal25/Starter-Notebooks/refs/heads/main/data/raw/content_refresh_anonymized.csv"

# Load the data directly from the web
df = pd.read_csv(url)

# --- Now calculate your real numbers ---
total_pages = len(df)
print(f"Number 1: There are {total_pages} pages in the dataset.")

# Check for the CTR column (it might be capitalized in the real file!)
# Let's make the code smart enough to find the right column name
ctr_col = 'ctr' if 'ctr' in df.columns else 'CTR'

if ctr_col in df.columns:
    low_ctr_pages = len(df[df[ctr_col] < 0.02])
    percentage_low = (low_ctr_pages / total_pages) * 100
    print(f"Number 2: {low_ctr_pages} pages ({percentage_low:.1f}%) have a CTR below 2%.")
    print(f"Number 3: The median CTR is {df[ctr_col].median():.4f}")
else:
    print("Could not find a CTR column. Here are the columns available:")
    print(df.columns.tolist())
```
